```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

X=df.drop(['MEDV'],axis=1)
y=df['MEDV']
columns= X.columns
scl=StandardScaler()
X=scl.fit_transform(X)
X=pd.DataFrame(X,columns=columns)
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.3)
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
class neural_network:
  def __init__(self,input_neurons,hidden_neurons,output_neurons,learning_rate):
        self.weights_input_hidden = np.random.uniform(-0.5, 0.5, (hidden_neurons, input_neurons))
        self.weights_output_hidden = np.random.uniform(-0.5, 0.5, (output_neurons, hidden_neurons))
        self.lr=learning_rate

  def activation(self,x):
    return 1/(1+np.exp(-x))

  def Forward(self, inputs):
    self.inputs = np.array(inputs, ndmin=2).T

    hidden_inputs = np.dot(self.weights_input_hidden, self.inputs)
    self.hidden_outputs = self.activation(hidden_inputs)

    final_inputs = np.dot(self.weights_output_hidden, self.hidden_outputs)
    return final_inputs

  def backward(self, inputs, targets, predictions):
    targets = np.array(targets, ndmin=2).T

    output_error = targets - predictions

    hidden_grad = np.dot(self.weights_output_hidden.T, output_error) * \
                  (self.hidden_outputs * (1 - self.hidden_outputs))

    self.weights_output_hidden += self.lr * np.dot(output_error, self.hidden_outputs.T)
    self.weights_input_hidden += self.lr * np.dot(hidden_grad, self.inputs.T)

  def MSE(self,targets,predictions):
    return np.mean((targets-predictions)**2)/len(targets)


                   
```

```python
input_neurons=X_train.shape[1]
hidden_neurons=32
output_neurons=1
learning_rate=0.01
epochs = 2000

model= neural_network(input_neurons,hidden_neurons,output_neurons,learning_rate)

```

```python
import sys 
losses={'Train':[],'Valid':[]}

for epoch in range(epochs):
    batch = np.random.choice(X_train.index, size=8)

    for record, target in zip(X_train.loc[batch].values, y_train.loc[batch].values):
        preds = model.Forward(record)
        model.backward(record, target, preds)

    # predictions
    train_preds = np.array([model.Forward(x) for x in X_train.values])
    valid_preds = np.array([model.Forward(x) for x in X_test.values])

    train_loss = model.MSE(y_train.values.reshape(-1,1), train_preds)
    valid_loss = model.MSE(y_test.values.reshape(-1,1), valid_preds)

    sys.stdout.write("\r PROGRESS : " + str(int(100 * epoch / float(epochs))) +
                     "%.. Training loss: " + str(train_loss) +
                     " .. Validation loss: " + str(valid_loss))

    losses['Train'].append(train_loss)
    losses['Valid'].append(valid_loss)
```

```python
results=pd.DataFrame()
results['preds']=model.Forward(X_test)[0]
results['actual']=y_test.values
results
```

```python
y_pred = np.array([model.Forward(x) for x in X_test.values])

y_pred = y_pred.reshape(-1)
y_true = y_test.values
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,6))
plt.scatter(y_true, y_pred,color='yellow')

plt.xlabel("Actual Values")
plt.ylabel("Predicted Values")
plt.title("Actual vs Predicted")


plt.plot([y_true.min(), y_true.max()],
         [y_true.min(), y_true.max()])



plt.figure(figsize=(10,5))
plt.plot(y_true[:50], label="Actual",color = 'red')
plt.plot(y_pred[:50], label="Predicted")

plt.legend()
plt.title("Actual vs Predicted (first 50 samples)")
plt.show()
```

```python
from sklearn.metrics import r2_score, mean_absolute_error

print("R2 Score:", r2_score(y_true, y_pred))
print("MAE:", mean_absolute_error(y_true, y_pred))
```
